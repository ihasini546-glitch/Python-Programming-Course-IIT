```python
class Car():
    color = "red"
    brand = "toyota"
    c_type = "hatchback"

    def start(self):
        print("car starts")

    def stop(self):
        print("car stopped")

    def move(self):
        print("car moving")

my_car = Car()
my_car.start()
```

```python
class Car:
    def __init__(self, brand, model, c_type):
        # instance attributes
        self.brand = brand
        self.model = model
        self.c_type = c_type

    def start(self):
        print(f"{self.brand} {self.model} {self.c_type} is starting......lets go!")

# Instantiate and call
my_car1 = Car("byd", "atto3", "suv")
my_car2 = Car("nissan", "allison240", "sedan")

my_car1.start()
my_car2.start()
```

```python
class Student:
    def __init__(self, name, age, grade):
        # instance attributes
        self.name = name
        self.age = age
        self.grade = grade

    def start(self):
        print(f"hello {self.name}! you are {self.age} yrs old and your grade is {self.grade}")

# Instantiate and call
student_01 = Student("hasini", "23", "a")
student_02 = Student("achini", "56", "b")

student_01.start()
student_02.start()
```
